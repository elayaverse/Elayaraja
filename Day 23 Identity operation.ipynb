```python
# Identity Operation (is, Is not)
# Used to Compare Memory location of two objects
```

```python
#Numeric Variable ( Same a and b numeric value)
a=10
b=10
print(id(a))
print(id(b))
a is b
```

```python
#( different a and b numeric value)
a=10
b=15
print(id(a))
print(id(b))
a is b
```

```python
# String
a=("python")
b=("python")
print(id(a))
print(id(b))
a is b
```

```python
a=("python")
b=("PYthon")
print(id(a))
print(id(b))
a is b
```

```python
#list 
a=[10,20,"hi"]
b=[10,20,"hi"]
print(id(a))
print(id(b))
a is b
```

```python
a=[10,20,"hi"]
b=[10,20,"hi"]
print(id(a))
print(id(b))
a is not b
```

```python
#Tuple
a=(15,25,"hi")
b=(15,25,"hi")
print(id(a))
print(id(b))
a is b
```

```python
a=(15,25,"hi")
b=(15,25,"hi")
print(id(a))
print(id(b))
a is not b
```

```python
# Dictionary
a={15:100,25:200,"E":"hi"}
b={15:100,25:200,"E":"hi"}
print(id(a))
print(id(b))
a is b
```

```python
a={15:100,25:200,"E":"hi"}
b={15:100,25:200,"E":"hi"}
print(id(a))
print(id(b))
a is not b
```

```python
# set
a={15,100,25,200,"E","hi"}
b={15,100,25,200,"E","hi"}
print(id(a))
print(id(b))
a is b
```
