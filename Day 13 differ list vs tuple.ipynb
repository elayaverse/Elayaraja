```python
# Syntax Differend (find type)
#list

l1=["venmozhi",6,4.3]
print(l1)
print(type(l1))
```

```python
#tuple

t1=("venmozhi",6,4.3)
print(t1)
print(type(t1))
```

```python
#mutable and immutable

# list is a mutable (can edit)
l2=[10,20,30,40]
l2[0]="hai"
print(l2)
```

```python
# tuple is a immutable (can not edit)
t2=(10,20,30,40)
t2[0]="hai"
print(t2)
```

```python
# deleting

# list can delete
l3=[10,20,30,40]
del l3[0]
print(l3)
```

```python
# tuple can not delete
t3=(10,20,30,40)
del t3[0]
print(t3)
```

```python
# Size comparision -- list vs tuple

#list is take more space
#tuple is take less space

l5=["rock","happy","learning"]
print(l5.__sizeof__())

t5=("rock","happy","learning")
print(t5.__sizeof__())
```
