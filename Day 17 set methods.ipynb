```python
# union() it join all one time word
a={"orange","yellow","black","pink"}
b={"pink","white","grey","green","black"}
a.union(b)
```

```python
a={"orange","yellow","black","pink"}
b={"pink","white","grey","green","black"}
c={"red"}
a.union(b,c)
```

```python
# intersection() ( find comman ibn both)
a={"orange","yellow","black","pink"}
b={"pink","white","grey","green","black"}
a.intersection(b)
```

```python
# intersection_update() ( it is update comman in other code)
a={"orange","yellow","black","pink"}
b={"pink","white","grey","green","black"}
a.intersection_update(b)
print(a)
```

```python
# difference()
a={"orange","yellow","black","pink"}
b={"pink","white","grey","green","black"}
a.difference(b)
```

```python
a={"orange","yellow","black","pink"}
b={"pink","white","grey","green","black"}
b.difference(a)
```

```python
# difference_update()
a={"orange","yellow","black","pink"}
b={"pink","white","grey","green","black"}
a.difference_update(b)
print(a)
```

```python
# symmetric_difference()
a={"orange","yellow","black","pink"}
b={"pink","white","grey","green","black"}
a.symmetric_difference(b)
```
