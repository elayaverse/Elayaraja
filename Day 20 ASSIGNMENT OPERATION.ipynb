```python
# Assignment Operation (=' +=, -=, *=, /=, //=, **=)
# To assign value to a variable 
```

```python
# aasignment ex
a=10
b=20
print(a,b)
a=b
print(a,b)
```

```python
a=10
b=20
print(a,b)
a=a+b
print(a,b)
```

```python
# additional assign
a=10
b=20
print(a,b)
a+=b
print(a,b)
```

```python
a=10
b=20
print(a,b)
a+=10
print(a,b)
```

```python
a=10
b=20
c=0
print(a,b,c)
c+=a+b
print(a,b,c)
```

```python
#Subtract assign
a=10
b=20
print(a,b)
a-=b
print(a,b)
```

```python
a=10
b=20
c=0
print(a,b,c)
c-=a+b
print(a,b,c)
```

```python
# multiple assign
a=10
b=20
print(a,b)
a*=b
print(a,b)
```

```python
#divide assign
a=20
b=10
print(a,b)
a/=b
print(a,b)
```

```python
# floor division
a=20
b=10
print(a,b)
a//=b
print(a,b)
```

```python
#moudulo assign
a=20
b=10
print(a,b)
a%=b
print(a,b)
```

```python
#exponential assign
a=2
b=3
print(a,b)
a**=b
print(a,b)
```
