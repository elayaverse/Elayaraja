```python
# Logical OR Operator

#Logical Operators runs with truth dable
                # OR truth Table is 
            #  Condition1            Condition2           Result
            #    True                  True                True
            #    True                  False               True
            #    False                 True                True
            #    False                 False               False
```

```python
# Applying the Table

x=10
(x<12)or(x>8)
```
