```python
# Dictionary 
# It is a mutable (can edit) data type 

#{name:"vanmozhi"}
#{key:value}
```

```python
# creat and find dictionary
d={}
type(d)
```

```python
d1={"name":"venmozhi","age":6}
print(d1)
```

```python
# list as dictionary ( covert list to dictionary)

a=["name","age"]
b=["venmozhi",6]
c=dict(zip(a,b))
print(c)
```

```python
# Tuple as dictionary ( covert tuple to dictionary)

a=("name","age")
b=("venmozhi",6)
c=dict(zip(a,b))
print(c)
```

```python
# list & tuple as dictionary ( covert tuple & list to dictionary)

a=("name","age")
b=["venmozhi",6]
c=dict(zip(a,b))
print(c)
```

```python
# dict() functions (key can incert with out double quote)

data=dict(name="venmozhi",age=6)
print(data)
```

```python
data=dict([("name","venmozhi"),("age",6)])
print(data)
```

```python
# lenght- basic operation
le={'name':'venmozhi','age':6}
len(le)
```

```python
#membership operation
le={'name':'venmozhi','age':6}
'age'in le
```
