```python
# adding elements= add()
s1={12,23,5.7,90,888}
s1 .add(100)
print(s1)
```

```python
# adding multiple elements= update() tuple
s1={12,23,5.7,90,888}
s1 .update(("python",100,"Elaya"))
print(s1)
```

```python
# adding multiple elements= update() list
s1={12,23,5.7,90,888}
s1 .update(["python",100,"Elaya"])
print(s1)
```

```python
# adding multiple elements= update() set
s1={12,23,5.7,90,888}
s1 .update({"python",100,"Elaya"})
print(s1)
```

```python
# adding multiple elements= update() set
s1={12,23,5.7,90,888}
s2={"python",100,"Elaya"}
s3=(s1,s2)
print(s3)
```

```python
#accessing elements in set- for() loop
s1={1,2,3}
for x in s1:
    print(x)
```

```python
#Remove() remove on elements only
f1={"elaya","python",12,34,56,889,98}
f1 .remove("python")
print(f1)
```

```python
#discard()
f1={"elaya","python",12,34,56,889,98}
f1 .discard("elaya")
print(f1)
```

```python
#discard()
f1={"elaya","python",12,34,56,889,98}
f1 .discard("elaya")
f1 .discard(889)
print(f1)
```
