```python
# string concatenation (with in the quotes called as concatenation)

firstname="Elayaraja"
lastname="G"
fullname=firstname+lastname
print(fullname)
```

```python
# space In between first and last name

firstname="Elayaraja"
lastname="G"
fullname=firstname+" "+lastname
print(fullname)
```

```python
firstname="elayaraja"
lastname="G"
fullname=firstname,lastname
print(fullname)
```

```python
# multiple text result

name="Elayaraja"
fullname=name*3
print(fullname)
```

```python
# printing statement in strings

a="learning python"
b="rock"
print("you are "+a+" in " +b)
```

```python
a="learning python"
b="rock"
print("you are {a} in {b}".format(a=a,b=b))
```

```python
a="learning python"
b="rock"
print("you are {} in {}".format(a,b))
```

```python
# string indexing (assign the number to string)
            # 0 1 2 3 4 
           #  h e l l o

a="hello"
a[0],a[1],a[2],a[3],a[4]
```

```python
# Negative string indexing (assign the number to string)
            # -5 -4 -3 -2 -1 
           #  h   e  l  l  o

a="hello"
a[-5],a[-4],a[-3],a[-2],a[-1]
```
