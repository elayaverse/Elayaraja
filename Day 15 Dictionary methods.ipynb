```python
# items ()
d={"name":"venmozhi","age":6,"height":4.5}
d.items()
```

```python
# keys()
d={"name":"venmozhi","age":6,"height":4.5}
d.keys()
```

```python
#values()
d={"name":"venmozhi","age":6,"height":4.5}
d.values()
```

```python
#access-get()
d={"name":"venmozhi","age":6,"height":4.5}
d.get("name")
```

```python
d={"name":"venmozhi","age":6,"height":4.5}
d.get("address", "not found")
```

```python
# ubdate()
d={"name":"venmozhi","age":6,"height":4.5}
d.update(address="salem")
print(d)
```

```python
d={"name":"venmozhi","age":6,"height":4.5}
d1={"address":"salem"}
d.update(d1)
print(d)
```

```python
d={"name":"venmozhi","age":6,"height":4.5}
d1={"address":"salem"}
d.update(d1,father="elaya")
print(d)
```

```python
# copy ()

d={"name":"venmozhi","age":6,"height":4.5}
d1=d.copy()
print(d1)
```

```python
# fromkeys ()
name=("sakthi","siva")
val=("nil")
data={}
data.fromkeys(name,val)
```

```python
# setdefault()
d={"name":"venmozhi","age":6,"height":4.5}
d.setdefault("father","elaya")
print(d)
```

```python
# pop()- delete
d={"name":"venmozhi","age":6,"height":4.5}
d.pop("name")
print(d)
```

```python
# popitem()- it has delete only last item
d={"name":"venmozhi","age":6,"height":4.5}
d.popitem()
print(d)
```

```python
#del () delete
d={"name":"venmozhi","age":6,"height":4.5}
del d["name"]
print(d)
```

```python
d={"name":"venmozhi","age":6,"height":4.5}
del d
print(d)
```

```python
# clear() it is clear all code in dictionary
d={"name":"venmozhi","age":6,"height":4.5}
d.clear()
print(d)
```
