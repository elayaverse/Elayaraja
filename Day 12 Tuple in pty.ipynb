```python
# accessing tuple values

# positive Vaiue

value=("rock",("happy","learning"),"python")
print(value[0])
print(value[1][0])
print(value[1][1])
print(value[2])
```

```python
#negative value
value=("rock",("happy","learning"),"python")
print(value[-1])
print(value[-2][0])
print(value[-2][1])
print(value[-3])
```

```python
value=("rock",("happy","learning"),"python")
print(value[-3])
print(value[-2][0])
print(value[-2][1])
print(value[-1])
```

```python
#slicing
#positive slicing
value=("rock",("happy","learning"),"python")
value[0:2]
```

```python
value=("rock",("happy","learning"),"python")
value[0:]
```

```python
value=("rock",("happy","learning"),"python")
value[:]
```

```python
# negative slicing


value=("rock",("happy","learning"),"python")
value[-3:-1]
```

```python
value=("rock",("happy","learning"),"python")
value[-3:]
```

```python
value=("rock",("happy","learning"),"python")
value[:-1]
```

```python
# packing and unpacking in tuple

a=("venmozhi",6,4.3)
(name,age,height)=a
print(name,age,height)
```

```python
a=("venmozhi",6,4.3)
(name,age,height)=a
print(name,name,height)
```

```python
a=("venmozhi",6,4.3,"height")
(name,*age,height)=a
print(name,age,height)
```
